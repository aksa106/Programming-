* Write a function, Given a string, return a dictionary containing the frequency of each character.
* Rules

    Preserve the characters as normal lowercase strings.
    Do not use collections.Counter.
    Use a function.

```python
def char_frequency(text):

    frequency = {}

    for char in text:
        frequency[char] = 0 # dict[key] = value [loop appends in empty dict]

    print(frequency)

char_frequency("Apple")
        
    
```

```python
def char_freq(text):

    freq = {} # empty dict gets created

    for char in text:   # every char gets looped through
        freq[char] = freq.get(char,0) + 1  # every individual char gets looped and becoems one key
    return freq

char_freq("banana")

```

```python
## Count vowels


def count_vowels(text):

    d={}

    for char in text:
        if char in 'aeiou':
            d[char] = d.get(char, 0)+1  # a : 1 , 

    return d

count_vowels("aapple")
            
```
