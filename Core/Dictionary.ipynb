# Dictionary

```python
scores = {
        "Alice": 85,
        "Bob": 92,
        "Charlie": 78,
        "David": 95
}

def highest_score(scores):

    highest = 0
    winner = ""

    for name,score in scores.items():
        if score>highest:
            highest = score
            winner = name

    return f"Highest score : {highest}, Name of winner : {winner}"
highest_score(scores)
 

```

* dict → keys by default
* Keys
    * for name in scores:
    * print(name)     

* dict.values() → values
* Values
    * for score in scores.values():
    * print(score)


* dict.items() → key + value
* Keys + values
    * for name, score in scores.items():
    * print(name, score)
