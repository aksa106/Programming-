```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

def common_elements(set1, set2):
    c = set1.intersection(set2)

    return c
common_elements(set1, set2)
```

```python
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7}

def common_elements(set1, set2):

    set3 = set()
    for item in set1:
        if item in set2:
            set3.add(item)

    return set3

common_elements(set1, set2)


## Unordered does not means if a thing is un iterable , python can still visitis
```
