# Recsursion

```python
def count(n):
    print(n)

    if n > 1:
        count(n-1)


count(3)
```

### Write a recursive function sum_n(n) that calculates the sum of numbers from 1 to n.

```python
def sum(n):

    if n == 0:
        return 0
        

    return n + sum(n-1)


print(sum(5))
```

```python
def multiply(n):

    if n == 0:
        return 1

    return n * multiply(n-1)


print(multiply(5))
```

### Write a recursive function power(n) that calculates 2 raised to the power n

```python
def power(n):
    if n == 0:
        return 1
        
    return 2 *power(n-1)

power(3)
```

```python
def gcd(m,n):
    (a,b) = (max(m,n), min(m,n))

    if a % b == 0:
        return(b)

    else:
        return(gcd(b, a%b))


print(gcd(24,130))
```

```python
def func(n):

    if n == 0:
        return 0

    return (n%10) + func(n//100)


func(12345)
```

```python
def fun(n):

    if n == 0:
        return 0

    if (n % 10) % 2 == 0:

        return 1 + fun(n // 10)

    else:

        return fun(n // 10)

fun(1234)
```
