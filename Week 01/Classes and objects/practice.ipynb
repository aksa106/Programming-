## Classes and Objects

### What is a class?

### Why do we need a class?
* Suppose you want to represent multiple students in a program. Each student has information , such as thier age. 

* Instead of writing separate code for every student you define  a common structure called class .

* here, student is class



```python
class Student: 
    def __init__(self, age): # init method initializes the object with info provided
        self.age = age # age is the parameter

        # self = refers to the current object
        # self.age is an attribute belonging to that object.
        # age is the parameter containing the supplied value 
        # = assigns the value to the attribute

        #For example, if the supplied age is 20, the assignment stores 20 in that 
        # object's age attribute.

s1 = Student(20)

#s1
# └── age: 20

#Python creates another Student object, this time initialized with the age 25.
s2 = Student(25)


print(s1.age)
print(s2.age)

```

    Both objects belong to the same class, but they store different values.

    This is the fundamental relationship between classes and objects:

    Student is the class.
    s1 and s2 refer to objects created from that class.
    age is an attribute of each object.
