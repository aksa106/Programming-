    Write a Python program that:
    1. Starts with x = 1.
    2. Uses a while loop to print consecutive positive integers.
    3. Stops when x becomes divisible by 5.
    4. Does not print the number divisible by 5.
    5. Prints all numbers on the same line, separated by spaces.

```python
x = 1

while True : 
    if x % 5 == 0:
        break
    print(x, end=' ')

    x += 1
```
