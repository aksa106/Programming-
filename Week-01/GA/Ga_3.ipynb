#### Write a Python function f(n) that takes a positive integer n as input and returns the number of odd factors of n, excluding 1 and n itself.

```python
def func(n):

    factorList = []
    sum = 0

    for i in range(1,n+1):
        if n%i == 0:

            factorList.append(i)

            sum += i

            

    return f" Factors : {factorList}, Sum : {sum} "



func(12)
```
