#### Write a Python function count_unique(S) that counts the number of distinct characters in a string, ignoring uppercase/lowercase differences.

```python
def count_unique(s):

    count = 0

    s = s.lower() #converted string intoe lowercase

    for i in range(len(s)):
        if s[i] not in s[:i]: ## s[i]  not in s [:i]
            count += 1

    return count


print(count_unique("apple"))


## s[:i] = s[0:i] means take charachters from the beginning of the string and upto index i , but NOT includin i.
```
