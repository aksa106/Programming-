## Check whether a list is a palindrome or not 

```python
def palindrome(L):

    if L == L[::-1]:
        return True

    return False

palindrome([1,2,3,3,2,1])


```

The .pop() method removes and returns an element from a list.

```python
def palindrome(L):

    while len(L) > 0:

        if L.pop(0) != L.pop(-1):

            return False
    return True 


palindrome([1,2,3,3,2,1])
```
