# Class vs Object ⭐
Class	Object
Blueprint	Instance of a class
Defines attributes and methods	Has attributes and can use methods
Created using class	Created by calling the class
Example: Student	Example: s1 = Student()


🧠 Remember
Class = Blueprint

Object = Actual thing

Attribute = Property

Method = Function inside a class

self = Current object

* Create a class called Student, create an object, and print the student's name.

```python
class Student:  ## defines a class named student 
    name = "Riya"


s1 = Student() ## creates an object instance of student's class

print(s1.name) ## Access the name attribute through the object
```

```python
# Create a class called Car that:
# 1. Has an attribute brand = "Toyota".
# 2. Creates an object called car1.
# 3. Prints the brand using that object.


class Car:
    brand = "Toyota"


car1 = Car()

print(car1.brand)

```
