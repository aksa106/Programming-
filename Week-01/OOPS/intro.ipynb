# Class vs Object ⭐
Class	Object
Blueprint	Instance of a class
Defines attributes and methods	Has attributes and can use methods
Created using class	Created by calling the class
Example: Student	Example: s1 = Student()


🧠 Remember
Class = Blueprint

Object = Actual thing

Attribute = Property

Method = Function inside a class

self = Current object

* Create a class called Student, create an object, and print the student's name.

```python
class Student:  ## defines a class named student 
    name = "Riya"


s1 = Student() ## creates an object instance of student's class

print(s1.name) ## Access the name attribute through the object
```

```python
# Create a class called Car that:
# 1. Has an attribute brand = "Toyota".
# 2. Creates an object called car1.
# 3. Prints the brand using that object.


class Car:
    brand = "Toyota"


car1 = Car()

print(car1.brand)

```

```python
# Multiple Objects

class Car :
    brand = "Toyota"

car1 = Car()

car2 = Car()

print(car1.brand)

print(car2.brand)
```

```python
class  Student:

    name = "Riya"


student1 = Student()

student2 = Student()


student2.name = "Arun"


print(student1.name)

print(student2.name)
```

## __init__() and __self__

So far, every object initially had the same name. But what if we want to give each student a different name when creating the object?

We use __init__().

```python
class Student:
    def __init__(self,name): 
        self.name = name 

## init : automatically runs when a object is created 

## self --> refers to current object 

## name : value passed when creating an object

##  self.name : Stores that value in the object.

s1 = Student("Arun")

s2 = Student("Arjun")

print(s1.name)
print(s2.name)
```

```python
class Car:
    def __init__(self,brand):

        self.brand = brand 

car1 = Car("BMW")

car2 = Car("Toyota")


print(car1.brand)

print(car2.brand)
```
