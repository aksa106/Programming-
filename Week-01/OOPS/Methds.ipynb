## Methods

### A method is a function defined inside a class.


```python
class flower:

    def __init__(self,name):

        self.name = name 

    def rose(self):
        print(self.name, "are beautiful")


flower1 = flower("Tuplips")

flower1.rose()
```

```python
class Cats :
    def __init__(self, name):

        self.name = name 

    def meow(self):

        print(self.name, "says meow")

cat1 = Cats("Kitty")

cat1.meow()
```

```python
class Enrollment:

    count = 0

    def __init__(self,n,c):
        self.name = n 
        self.course = c
        Enrollment.count += 1 

    def display(self):

        print(f"Name : {self.name} , Course : {self.course}")
       


e1 = Enrollment("Python", "abc")
e2 = Enrollment("xyz", "maths")


e1.display()
e2.display()

print(f"Enrollment_count : {Enrollment.count}")
```
