## Methods

### A method is a function defined inside a class.


```python
class flower:

    def __init__(self,name):

        self.name = name 

    def rose(self):
        print(self.name, "are beautiful")


flower1 = flower("Tuplips")

flower1.rose()
```

```python
class Cats :
    def __init__(self, name):

        self.name = name 

    def meow(self):

        print(self.name, "says meow")

cat1 = Cats("Kitty")

cat1.meow()
```
