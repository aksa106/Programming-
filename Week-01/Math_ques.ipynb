```python
def factors(n):

    factorlist = []

    for i in range(1,n+1):

        if n%i == 0 :

            factorlist.append(i)

    return factorlist

print(factors(12))
```

###  Checking Whether a Number Is Prime
* A prime number has exactly two positive factors: 1 and itself.

```python
def prime(n):

    if n < 2:
        return False 
    
    for i in range(2,n):

        if n % i == 0:

            return False 

    return True 


prime(7)



```

* Print prime numbers from 1 to 100 


```python
def prime(n):

    if n<2:
        return False 

    for i in range(2,n):
        if n % i == 0:
            return False 

    return True 


def primeList():

    primeList = []

    for i in range(2,101):
        if prime(i):
            primeList.append(i)

    return primeList


print(primeList())
```

* GCD (Greatest Common Divisor)

    * The GCD of two numbers is the largest positive number that divides both exactly.

    Factors of 12: 1, 2, 3, 4, 6, 12

    Factors of 18: 1, 2, 3, 6, 9, 18

    Common factors: 1, 2, 3, 6

    GCD = 6 ✅
